```python
import pandas as pd
import plotly.express as px
import requests
import io

# 1. URL do seu arquivo
url = "https://raw.githubusercontent.com/103658-beep/Bilu_soq_original/refs/heads/main/data.csv"

# Carregando os dados de forma segura
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

if response.status_code == 200:
    df = pd.read_csv(io.StringIO(response.text))

    # 2. Agrupando os dados por hora (ou data) para encontrar a Mínima e a Máxima
    # Caso prefira agrupar por minuto, mude 'hora' para a coluna desejada
    df_agrupado = df.groupby('hora')['temperatura_C'].agg(['min', 'max']).reset_index() if 'hora' in df.columns else df.agg({'temperatura_C': ['min', 'max']})

    # Se o agrupamento acima gerou um formato específico, vamos garantir que temos as colunas limpas:
    # Renomeando para facilitar a leitura
    df_agrupado = df.groupby('hora').agg(
        Temperatura_Minima=('temperatura_C', 'min'),
        Temperatura_Maxima=('temperatura_C', 'max')
    ).reset_index()

    # Pegando apenas as primeiras 15 linhas para o gráfico não ficar poluído
    df_plot = df_agrupado.head(15)

    # 3. Transformando o DataFrame para o formato ideal do Plotly (formato longo)
    df_melted = df_plot.melt(
        id_vars=['hora'],
        value_vars=['Temperatura_Minima', 'Temperatura_Maxima'],
        var_name='Métrica',
        value_name='Temperatura'
    )

    # Renomeando as legendas para ficar mais bonito
    df_melted['Métrica'] = df_melted['Métrica'].replace({
        'Temperatura_Minima': 'Mínima (°C)',
        'Temperatura_Maxima': 'Máxima (°C)'
    })

    # 4. Criando o gráfico de barras agrupadas
    fig = px.bar(
        df_melted,
        x='hora',
        y='Temperatura',
        color='Métrica',
        barmode='group', # Deixa as barras lado a lado (Mínima vs Máxima)
        title='<b>Temperatura Mínima e Máxima por Horário</b>',
        text_auto='.1f', # Mostra o valor com 1 casa decimal em cima da barra
        color_discrete_map={
            'Mínima (°C)': '#3498db', # Azul para mínima
            'Máxima (°C)': '#e74c3c'  # Vermelho para máxima
        }
    )

    # 5. Personalizando o design
    fig.update_layout(
        plot_bgcolor='white',
        paper_bgcolor='white',
        font=dict(family='sans-serif', size=14, color='#333333'),
        title_font_size=18,
        xaxis=dict(showgrid=False, title='Horário'),
        yaxis=dict(showgrid=True, gridcolor='#eaeaea', title='Temperatura (°C)'),
        legend=dict(title='Legenda', orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1),
        margin=dict(t=80, b=50, l=50, r=50)
    )

    # Exibe o gráfico interativo
    fig.show()

else:
    print(f"Erro ao baixar o arquivo. Código: {response.status_code}")
```
